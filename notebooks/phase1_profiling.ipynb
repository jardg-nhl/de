```python
from pathlib import Path
import json
import pandas as pd

profile_path = Path('lakehouse/reports/profile.json')
profile = json.loads(profile_path.read_text(encoding='utf-8'))
pd.DataFrame([{'file': row['file'], 'rows': row['rows'], 'empty_cells': row['empty_cells']} for row in profile])
```

```python
rating = next(row for row in profile if row['file'] == 'rating.csv')
{'rating_distribution': rating['rating_distribution'],
 'user_movie_key_order_nondecreasing': rating['user_movie_key_order_nondecreasing'],
 'duplicate_user_movie_pairs': rating['duplicate_user_movie_pairs'],
 'event_time_min': rating['event_time_min'],
 'event_time_max': rating['event_time_max'],
 'max_ratings_per_user': rating['user_rating_max'],
 'median_ratings_per_user': rating['user_rating_median']}
```
